```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.metrics import f1_score,mean_squared_error,confusion_matrix
from sklearn.tree import DecisionTreeClassifier
```

```python
df= pd.read_csv("shop_smart_ecommerce.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = pd.get_dummies(
    data=df,
    columns=["VisitorType", "Month"],
    drop_first=True
)
```

```python
X=df.drop(columns=["Revenue"])
Y=df["Revenue"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,random_state=42,test_size=0.2)
```

```python
model=DecisionTreeClassifier()
model.fit(x_train,y_train)
y_ped=model.predict(x_test)
model.score(x_test,y_test)
```

```python
print("F1 Score: ",f1_score(y_ped,y_test))
print("Mean_squared_error Score: ",mean_squared_error(y_ped,y_test))
confusion_matrix(y_ped,y_test)
```

```python
path=model.cost_complexity_pruning_path(x_train,y_train)
ccp_alphas=path.ccp_alphas
print(ccp_alphas)
```

```python
best_acc=0
best_model=0
best_alpha=0
for alpha in ccp_alphas:
    model=DecisionTreeClassifier(
        # max_depth=10,
        # min_samples_split=5,
        random_state=42,
        ccp_alpha=alpha
    )
    model.fit(x_train,y_train)
    acc=model.score(x_test,y_test)
    if acc>best_acc:
        best_acc=acc
        best_alpha=alpha
        best_model=model
print(best_acc)      
```

```python
y_ped=best_model.predict(x_test)
print("F1 Score: ",f1_score(y_ped,y_test))
print("Mean_squared_error Score: ",mean_squared_error(y_ped,y_test))
confusion_matrix(y_ped,y_test)
```
